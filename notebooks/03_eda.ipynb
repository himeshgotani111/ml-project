```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style for clean, modern plots
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["font.size"] = 11
```

```python
df = pd.read_csv("../dataset/processed/cardio_cleaned.csv")
print("Cleaned Dataset Shape:", df.shape)
df.head()
```

```python
plt.figure(figsize=(6, 4))
ax = sns.countplot(data=df, x="cardio", palette=["#2ecc71", "#e74c3c"])
plt.title("Cardiovascular Disease Distribution (0 = Healthy, 1 = Disease)")
plt.xlabel("Target (cardio)")
plt.ylabel("Count")

# Add count annotations above each bar
for p in ax.patches:
    ax.annotate(f'{int(p.get_height()):,}', (p.get_x() + 0.25, p.get_height() + 500))

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(data=df, x="age_years", hue="cardio", multiple="dodge", bins=25, 
             palette=["#2ecc71", "#e74c3c"], kde=True)
plt.title("Age Distribution by Cardiovascular Disease Status")
plt.xlabel("Age (Years)")
plt.ylabel("Number of Patients")
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Systolic BP (ap_hi)
sns.boxplot(ax=axes[0], data=df, x="cardio", y="ap_hi", palette=["#2ecc71", "#e74c3c"])
axes[0].set_title("Systolic Blood Pressure (ap_hi) vs Disease")
axes[0].set_xlabel("Cardio (0 = No, 1 = Yes)")
axes[0].set_ylabel("Systolic BP (mmHg)")

# Diastolic BP (ap_lo)
sns.boxplot(ax=axes[1], data=df, x="cardio", y="ap_lo", palette=["#2ecc71", "#e74c3c"])
axes[1].set_title("Diastolic Blood Pressure (ap_lo) vs Disease")
axes[1].set_xlabel("Cardio (0 = No, 1 = Yes)")
axes[1].set_ylabel("Diastolic BP (mmHg)")

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Cholesterol vs Disease
sns.barplot(ax=axes[0], data=df, x="cholesterol", y="cardio", palette="Reds", errorbar=None)
axes[0].set_title("Disease Rate by Cholesterol Level")
axes[0].set_xlabel("Cholesterol (1: Normal, 2: Above Normal, 3: High)")
axes[0].set_ylabel("Probability of Disease")

# Glucose vs Disease
sns.barplot(ax=axes[1], data=df, x="gluc", y="cardio", palette="Blues", errorbar=None)
axes[1].set_title("Disease Rate by Glucose Level")
axes[1].set_xlabel("Glucose (1: Normal, 2: Above Normal, 3: High)")
axes[1].set_ylabel("Probability of Disease")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.boxplot(data=df, x="cardio", y="bmi", palette=["#2ecc71", "#e74c3c"])
plt.title("Body Mass Index (BMI) vs Cardiovascular Disease")
plt.xlabel("Cardio (0 = No, 1 = Yes)")
plt.ylabel("BMI (kg/m²)")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 8))

# Calculate Pearson correlation matrix
corr_matrix = df.corr()

# Plot heatmap
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", cbar=True, linewidths=0.5)
plt.title("Feature Correlation Heatmap", fontsize=14)
plt.tight_layout()
plt.show()
```
